```python
#==================== UTILIZADO PARA "CHAMAR" AS BIBLIOTECAS 
#==================== NECESSÁRIAS DE DENTRO DO PYTHON

import duckdb
import pandas as pd
import os
from datetime import datetime
```

```python
#==================== CONECTANDO COM O BANCO DE DADOS DUCKDB

con = duckdb.connect(database='dados_duckdb.db', read_only=False)
```

```python
#==================== LENDO O ARQUIVO CSV

arquivo = 'z0019_2.csv' #nome do arquivo CSV
data_ingestao = datetime.now() 
df = pd.read_csv(f'../landing/{arquivo}', sep=';')
df['nome_arquivo'] = arquivo
df['data_ingestao'] = data_ingestao

df.head()
```

```python
#con.execute("DROP TABLE IF EXISTS bronze_produtos")
```

```python
#=================== CRIANDO TABBELA PARA A INSERÇÃO DOS DADOS
con.execute("""
CREATE TABLE IF NOT EXISTS bronze_produtos (
    NATBR VARCHAR,
    MAKTX VARCHAR,
    WERKS VARCHAR,
    MAINS VARCHAR,
    LABST VARCHAR,
    nome_arquivo VARCHAR,
    data_ingestao VARCHAR
    )
""")
```

```python
#======================== INSERINDO DADOS NA TABELA CRIADA
con.execute("INSERT INTO bronze_produtos SELECT * from df")
```
